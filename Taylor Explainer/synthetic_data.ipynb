# Data with numeric features

```python
X= synthetic.data_generator_4ft(n=200,df=True)
X
```

```python
mean_n1= np.array([-2,2.0])
mean_n2= np.array([0.0,0.0])

D_n1= np.array([[1.0,0.0],[0.0,1.0]]).T
D_n2= np.array([[1.0,1.0],[-1.0,1.0]]).T

S_n1= np.diag([2.0,1.0])
S_n2= np.diag([4.0,0.1])

C_n1= np.linalg.multi_dot([D_n1,S_n1,D_n1.T])
C_n2= np.linalg.multi_dot([D_n2,S_n2,D_n2.T])

X1= np.random.multivariate_normal(mean_n1,C_n1,size=(10))
X2= np.random.multivariate_normal(mean_n2,C_n2,size=(10))

X= np.zeros((X1.shape[0] + X2.shape[0],5))
```

```python
X_plot= synthetic.data_generator_4ft(200)

# Creating figure and axis objects using subplots()
fig, ax= plt.subplots(figsize=[7, 5])

ax.set_xlabel('core_1', fontsize=13)
ax.set_ylabel('core_2', fontsize=13)

plt.scatter(X_plot[:,0],X_plot[:,1],c=X_plot[:,-1])

plt.savefig('synt_ds.svg')
```
