```python
import openml
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

# Adult

```python
dataset = openml.datasets.get_dataset(1590)
X, y, _, _ = dataset.get_data(dataset_format="dataframe")
y = X[['class']]
X = X.drop(['class'], axis=1)
y = LabelEncoder().fit_transform(y.values.ravel())
X['target'] = y
X.to_csv('./data/adult/data.csv', index=False)
```

# Compas

```python
dataset = openml.datasets.get_dataset(42192)
X, y, _, _ = dataset.get_data(dataset_format="dataframe")
y = X[['two_year_recid']]
X = X.drop(['two_year_recid'], axis=1)
X['target'] = y
X.to_csv('./data/compas/data.csv', index=False)
```
